1. Write a Python program to check that a string contains only a certain set of characters (in this case a-z, A-Z and 0-9).

```python
import re
```

```python
text = input("Enter a string: ")
regex = r'([a-zA-Z0-9]+)'
if re.fullmatch(regex, text):
    print("Valid input")
else:
    print("Invalid input")
```

```python
text = input("Enter a string: ")
regex = r'(\w+)'
if re.fullmatch(regex, text):
    print("Valid input")
else:
    print("Invalid input")
```

2. Write a Python program that matches a string that has an ‘a’ followed by anything, ending in ‘b’.

```python
text = input("Enter text: ")
regex = r'^a.*b$'
if re.fullmatch(regex, text):
    print("Valid input")
else:
    print("Invalid input")
    
```

3. Write a program that uses regex to validate user input. Examples:
Check if a password meets certain criteria (e.g., minimum length, containing uppercase and lowercase letters).
Validate a date format (e.g., DD/MM/YYYY).

```python
# Password criteria
password = input("Enter password: ")
regex = r'^(\w+).{8,}$'
if re.fullmatch(regex, password):
    print("Valid password")
else:
    print("Invalid password")
```

```python
password = input("Enter password: ")
regex = r'^(?=.*[a-z])(?=.*[A-Z])(?=.*\d).{8,}$'
if re.fullmatch(regex, password):
    print("Valid password") 
else:
    print("Invalid password")
```

```python
# Date format
date = input("Enter date: ")
regex = r"\b\d{2}/\d{2}/\d{4}\b"
if re.fullmatch(regex, date):
    print("Valid date format")
else:
    print("Invalid date format")
```

4. Write a program that uses regex to extract specific information from a string. Examples:
Extract email addresses: Find all strings containing “@” and a dot (.).
Extract phone numbers: Find all strings with a specific format (e.g., XXX-XXX-XXXX).

```python
string = "My email address is johndoe@python.com and my phone number is 0701234567"
regex = r'\w+@\w+\.\w+'
matches = re.findall(regex, string)
print (matches) 
```

```python
# Extracting a phone number from a string
regex = r'\d+'
matches = re.findall(regex, string)
print (matches)
```

5. Write a program that uses regex to search for and replace text in a string. Example:
Replace all occurrences of “Mr.” with “Ms.”: text.replace(r”Mr.”, “Ms.”) (r prefix for raw string)

```python
string = "Mr. John Doe and Mr. Smith are related to Ms. Jane Doe and Mr. Alice Down"
text = re.sub(r'Mr\.', 'Ms.', string)
print(text)
```

```python
messy_data = """
Customer Name: Jane Doe, Email: janedoe@gmail.com, Phone: +254-712-345678
Name: John Smith; Email: john_smith@company.co.ke; Tel: 0712345678
Customer=Alice; Email=alice@my-mail.org, Contact=+254712345679
Username: mike.1990! Email --> mike1990@yahoo.com  Number: 0722 334 445
random text here, no useful data

Steve | steve@outlook.com | 0700-112-233 | Nairobi
"""
```

Extract Email addresses.

```python
emails = re.findall(r'\w+[\w.+-]*@\w+\.\w+', messy_data)
print("Emails:", emails)
```

Extract phone numbers

```python
numbers = re.findall(r'\+[\d\s-]{9,}|07\d[\d\s-]{7,}', messy_data)
print("Phone Numbers:", numbers)
```

Only names with letters

```python
names = re.findall(r'[A-Z][a-z]+\s[A-Z][a-z]+', messy_data)
print("Names:", names)
```

Non-word characters

```python
non_word_chars = re.findall(r'\W', messy_data)
print("Non-word characters:", set(non_word_chars))
```
