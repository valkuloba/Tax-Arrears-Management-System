```python
import requests
```

```python
# Make a GET request to a sample API
url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url)
response.status_code
```

```python
# Extract JSON data
posts = response.json()
posts
```

```python
# extracting titles from the JSON data
for post in posts:
    print(f"Title: {post['title']}")
    print(f"Body: {post['body']}")
    print()
```

```python
import csv
```

```python
with open('posts.csv', 'w') as file:
    writer = csv.writer(file)
    # Write header
    writer.writerow(['Title', 'Body'])
    # Write data
    for post in posts:
        writer.writerow([post['title'], post['body']])

print("Data written to posts.csv")
```

Get country information using REST countries API

```python
import requests
```

```python
country = input("Enter country name: ")
country
```

```python
#API request
url = f"https://restcountries.com/v3.1/name/{country}"
response = requests.get(url)
```

```python
response.status_code
```

```python
data = response.json()
data
```

```python
#Extract country data
country_name = data[0]['name']['common']
capital = data[0]['capital']
region = data[0]['region']
population = data[0]['population']
currencies = ', '.join([currencies[key]['name'] for key in currencies])
```

```python
print(f"\nCountry: {country_name}")
print(f"Capital: {capital}")    
print(f"Region: {region}")
print(f"Population: {population}")
print(f"The currency is {currencies}")
```
